# Environment check

Launch JupyterLab with the notebook dependency group and your PyTorch profile.
This notebook imports the shared project package and performs a tiny calculation.
It does not download datasets or weights. Clear all outputs before committing.

```python
import sys

import matplotlib
import nibabel as nib
import numpy as np
import pandas as pd
import scipy
import torch

from ml_biomed_project.devices import select_device

device = select_device()
print(
    {
        "NumPy": np.__version__,
        "SciPy": scipy.__version__,
        "pandas": pd.__version__,
        "Matplotlib": matplotlib.__version__,
        "NiBabel": nib.__version__,
    }
)
print(f"Python: {sys.executable}")
print(f"PyTorch: {torch.__version__}; device: {device}")
with torch.inference_mode():
    x = torch.ones((4, 4), device=device)
    result = (x @ x).cpu()
assert torch.equal(result, torch.full((4, 4), 4.0))
print("Calculation passed.")
```
